# IBCP Library Examples

This notebook demonstrates how to use the IBCP library to interact with the Interactive Brokers Client Portal API.


```python
import ibcp


# Initialize the REST client
api = ibcp.REST()

# Get account information
accounts = api.get_accounts()
print(f"Accounts: {accounts}")
```

```python
# Get cash balance
balance = api.get_cash_balance()
print(f"Cash balance: {balance}")
```
